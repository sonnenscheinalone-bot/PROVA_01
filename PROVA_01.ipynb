CONTROLE DE ENTRADA DE CINEMA
CONTEXTO: O programa deve validar a entrada de pessoas em uma sala de cinema com base na idade e no valor disponivel para a compra do ingresso.

ENTRADA - Perguntar ao usuario a idade e o valor total que ele tem na carteira.
DECISÃO - Se a idade for menor de 16 anos, exibir "entrada permitida apenas com acompanhante"
Se a idade for 16 ou mais, verificar o dinheiro disponivel:
Se tiver 30,00 ou mais: "Ingresso comprado! Bom filme"
Se tiver menos de 30,00: "Saldo insuficiente para o ingresso"
FINALIZAÇÃO - Mostre a mensagem "Foi bom te atender"

```python
idade = int(input("qual a sua idade?"))
if idade < 16:
    print("Entrada permitida apenas com acompanhante")
    if idade >= 16:
      print("Verificar dinheiro na carteira")
valor = float(input("Qual o valor tem na carteira?"))
if valor >= 30:
  print("Ingresso comprado! Bom filme")
if valor < 30:
  print("Saldo insuficiente para o ingresso")
  print("Foi bom te atender")
```
